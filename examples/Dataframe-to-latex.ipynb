```python
import pandas as pd
import numpy as np
from pythonlatex.pandas_to_latex import PandasTableFormatter
```

```python
df = pd.read_csv("ressources/mol_results_example.csv", index_col=0).reset_index(drop=True)
df
```

```python
formatter  = PandasTableFormatter(
    n_decimals = 3, # Number of decimals to keep in the table
    aggregation_methods=["mean", "std", "min", "max"], # Aggregation functions to apply to the data
    main_subset=0, # Subset of values to bold, here the first column will be bolded corresponding to the mean values, if [0,1] the first two columns will be bolded (independently)
)
```

```python
style = formatter.style(
    df, # Dataframe to format
    rows= "embedder", # Rows 
    cols=["category", "dataset"], # Columns 
    values= "val_AUROC", # Values
    highlight_fn= np.nanmax, # Function to use to highlight the values, here the maximum values will be highlighted
    props=["font-weight: bold; text-decoration: underline;", "font-weight: bold;"], # Properties to apply to the highlighted values, here the maximum values will be underlined and bolded, the second maximum values will be bolded
    hide_agg_labels=False, # Hide the aggregation column names in the latex
    special_format_agg = {
        "std": lambda x: "\\tiny $\\pm$" + x, # Format to apply to the standard deviation values
        "min": lambda x: "\\tiny ( "+ x + " - ", # Format to apply to the minimum values
        "max": lambda x: "\\tiny" + x + " )", # Format to apply to the maximum values
    }
)
style
```

```python
print(style)
```

```python
formatter.save_to_latex(style, "ressources/mol_results_example.tex", multicol_align="|c|", hrules=True)
```
