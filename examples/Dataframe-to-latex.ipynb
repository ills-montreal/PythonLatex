```python
import pandas as pd

from pythonlatex.pandas_to_latex import PandasTable
from pythonlatex.highlight_formatter import HighlightFormatter
from tests.test_pandas import highlight_format
```

```python
df = pd.read_csv("ressources/mol_results_example.csv", index_col=0).reset_index(drop=True)
df
```

# Example 1
We will create a table with the following features:

- We highlight the maximum values in the table
- We underline and bold the maximum values, bold the second maximum values


```python
formatter = HighlightFormatter(
    highlight_method = "max", # What values need to be highlighted (max, min, or a custom function)
    highlight_format = ["\\textbf{\\underline{", "\\textbf{"] # How to format the highlighted values bold, underlined, etc.
)

table = PandasTable(
    df,  # Dataframe to convert to format
    higlighter = formatter, # HighlightFormatter object or a string if "max" or "min", will bold the max or min values
    n_decimals = 3, # Number of decimals to keep in the table
    rotate = "", # Whether to rotate the name of the columns
    add_std = True, # Whether to add the standard deviation of the values
    multicols = None, # Whether to add multi-columns (see below for more details)
)
```

```python
table["embedder", "dataset", "val_AUROC"]
```

```python
table["embedder", "category", "val_AUROC"]
```

# Example 2
We will create a table with the following features:

- We highlight the maximum values in the table
- We underline and bold the maximum values, bold the second maximum values
- The category of the task will be added at the top

```python
# We specify the multi-columns to be added, we want for each dataset, the category, and the dataset name
multicols = {
    dataset: [df[df["dataset"] == dataset]["category"].unique()[0], dataset] for dataset in df["dataset"].unique()
}
multicols
```

```python
formatter = HighlightFormatter(
    highlight_method = "max", # What values need to be highlighted (max, min, or a custom function)
    highlight_format = ["\\textbf{\\underline{", "\\textbf{"] # How to format the highlighted values bold, underlined, etc.
)
table = PandasTable(
    df,  # Dataframe to convert to format
    higlighter = formatter, # HighlightFormatter object or a string if "max" or "min", will bold the max or min values
    n_decimals = 3, # Number of decimals to keep in the table
    rotate = "", # Whether to rotate the name of the columns
    add_std = True, # Whether to add the standard deviation of the values
    multicols = multicols, # Whether to add multi-columns (see below for more details)
)
```

```python
table["embedder", "dataset", "val_AUROC"]
```

```python
table.save_latex("ressources/mol_results_example.tex")
```
